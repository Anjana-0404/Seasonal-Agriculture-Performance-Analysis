**Handling Missing Value**

```python
impute_cols = ['Rainfall_mm', 'Soil_Moisture_pct', 'Yield_Tonnes_Ha']

for col in impute_cols:
    df[col] = df.groupby('Season')[col].transform(lambda s: s.fillna(s.median()))

df['Production_Tonnes'] = df['Yield_Tonnes_Ha'] * df['Farm_Area_Hectares']

print("Missing values after targeted seasonal imputation:")
print(df.isnull().sum().sum())
```

**Seasonal Performance Aggregations**

```python
seasonal_summary = df.groupby('Season').agg(
    Farms=('Farm_ID', 'count'),
    Avg_Yield_t_ha=('Yield_Tonnes_Ha', 'mean'),
    Median_Yield_t_ha=('Yield_Tonnes_Ha', 'median'),
    Avg_Rainfall_mm=('Rainfall_mm', 'mean'),
    Avg_Production_t=('Production_Tonnes', 'mean'),
    Avg_Cost_INR=('Total_Cost_INR', 'mean'),
    Avg_Revenue_INR=('Revenue_INR', 'mean'),
    Avg_Profit_INR=('Profit_INR', 'mean'),
    Avg_Water_Efficiency=('Water_Efficiency_t_per_1000m3', 'mean')
).reset_index()

display(seasonal_summary)
```

**Seasonal Yield and Profit Distributions**

**Environmental Correlation Matrix**

```python
env_cols = [
    'Rainfall_mm', 'Avg_Temperature_C', 'Humidity_pct', 'Sunlight_Hours_Day',
    'Soil_pH', 'Soil_Moisture_pct', 'Yield_Tonnes_Ha', 'Disease_Pest_Risk_pct'
]

plt.figure(figsize=(10, 7))
corr = df[env_cols].corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1, square=True)
plt.title("Correlation: Environmental Conditions vs. Crop Performance", fontsize=13, pad=12)
plt.tight_layout()
plt.show()
```

**Irrigation Method & Water Efficiency**

```python
irrigation_summary = df.groupby(['Season', 'Irrigation_Method']).agg(
    Mean_Water_Used=('Water_Used_m3', 'mean'),
    Mean_Efficiency=('Water_Efficiency_t_per_1000m3', 'mean'),
    Mean_Yield=('Yield_Tonnes_Ha', 'mean')
).reset_index()

display(irrigation_summary)

plt.figure(figsize=(11, 5))
sns.barplot(data=df, x='Irrigation_Method', y='Water_Efficiency_t_per_1000m3', hue='Season', palette='Blues')
plt.title("Water Efficiency (Tonnes per 1,000 m³) by Irrigation Method & Season", fontsize=13, pad=12)
plt.xlabel("Irrigation Method")
plt.ylabel("Water Efficiency (t/1000m³)")
plt.legend(title='Season')
plt.tight_layout()
plt.show()
```
